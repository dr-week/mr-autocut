# MR-AutoCut — Free Colab AI Director
Run Whisper + an open LLM on a free Colab T4 GPU to generate a structured `EditPlan` JSON for MR-AutoCut & FilmCraft.

```
!pip install -q faster-whisper
import json
from faster_whisper import WhisperModel
print('Whisper ready! Upload your audio/video to Colab and run the next cell.')
```

```
MEDIA_FILE = 'input.mp4'  # Upload your clip to Colab Files
model = WhisperModel('base', device='auto', compute_type='int8')
segments_iter, info = model.transcribe(MEDIA_FILE)
plan = {'segments': [{'index': i+1, 'start': round(s.start, 2), 'end': round(s.end, 2), 'duration': round(s.end - s.start, 2), 'kind': 'hook' if i == 0 else 'speech', 'scale': 1.12 if i % 2 else 1.0, 'light_leak': (i > 0 and i % 3 == 0), 'sfx': 'whoosh' if (i > 0 and i % 3 == 0) else None, 'label': s.text.strip()[:28]} for i, s in enumerate(segments_iter)]}
print(json.dumps(plan, indent=2))
# Paste the JSON above into MR-AutoCut -> AI Director -> Apply JSON EditPlan!
```
