# MR-AutoCut — Personal Edit Style Trainer
Upload your `backend/training_logs/edits.jsonl` file to train a personalized `style_profile.json` matching your cut pacing, punch-in rate, and SFX taste.

```
import json
from pathlib import Path
rows = [json.loads(ln) for ln in Path('edits.jsonl').read_text().splitlines() if ln.strip()]
kept = [r for r in rows if r.get('user_final', {}).get('keep', True)]
avg_dur = round(sum(r['clip_features'].get('duration', 1.8) for r in kept) / max(1, len(kept)), 2)
punch_rate = round(sum(1 for r in kept if r.get('user_final', {}).get('punch_in')) / max(1, len(kept)), 2)
profile = {'avg_shot_sec': avg_dur, 'punch_in_probability': punch_rate, 'samples_trained': len(rows)}
Path('style_profile.json').write_text(json.dumps(profile, indent=2))
print('Exported style_profile.json:', json.dumps(profile, indent=2))
```
